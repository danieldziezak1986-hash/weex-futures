# WEEX V6.44 — TOP 150 Crypto + TOP 50 TradFi
Uruchom komórkę poniżej. Przy pierwszym użyciu zaakceptuj dostęp do Dysku Google na potrzeby trwałej historii OI. ZIP z wynikami pobierze się automatycznie. Kolektor nie wysyła zleceń.

```python
import requests
import subprocess
import sys
from pathlib import Path
from google.colab import drive, files

# 1. Trwała historia OI na Dysku Google (wymaga autoryzacji przy pierwszym użyciu)
drive.mount("/content/drive")
state_path = Path("/content/drive/MyDrive/WEEX/weex_v644_oi_history_state.json")
state_path.parent.mkdir(parents=True, exist_ok=True)

# 2. Pobierz aktualny kolektor V6.44 z GitHub
url = "https://raw.githubusercontent.com/danieldziezak1986-hash/weex-futures/main/weex_v644_collector.py"
response = requests.get(url, timeout=30)
response.raise_for_status()
script_path = Path("/content/weex_v644_collector.py")
script_path.write_bytes(response.content)

# 3. Skan: TOP 150 Crypto + TOP 50 TradFi
output_dir = Path("/content/weex_v644_results")
output_dir.mkdir(parents=True, exist_ok=True)
command = [
    sys.executable, "-u", str(script_path),
    "--top", "150", "--tradfi-top", "50",
    "--max-deep", "30", "--state-path", str(state_path),
    "--out", str(output_dir)
]
subprocess.run(command, check=True)

# 4. Pobierz wynik ZIP automatycznie do swojego urządzenia
archives = list(output_dir.glob("weex_v644_*.zip"))
if not archives:
    raise FileNotFoundError("Nie znaleziono archiwum WEEX V6.44")
latest_zip = max(archives, key=lambda file: file.stat().st_mtime)
print("Pobieranie wyników:", latest_zip.name)
files.download(str(latest_zip))
```
