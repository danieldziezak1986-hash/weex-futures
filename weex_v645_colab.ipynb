# WEEX V6.45 — skaner z historią OI
TOP 150 Crypto, TOP 50 TradFi. Stan na Dysku Google, automatyczny ZIP. Żadnego handlu automatycznego.

```python
import requests
from pathlib import Path
from google.colab import drive, files

# Trwała historia OI na Dysku Google (jednorazowa zgoda)
drive.mount("/content/drive")
stan = Path("/content/drive/MyDrive/WEEX/weex_v645_oi_history_state.json")
stan.parent.mkdir(parents=True, exist_ok=True)

# Pobranie kolektora z GitHub
url = "https://raw.githubusercontent.com/danieldziezak1986-hash/weex-futures/main/weex_v645_collector.py"
r = requests.get(url, timeout=30)
r.raise_for_status()
with open("weex_v645_collector.py", "wb") as f:
    f.write(r.content)

# TOP 150 Crypto + TOP 50 TradFi
%run weex_v645_collector.py --top 150 --tradfi-top 50 --max-deep 30 --state-path $stan

# Automatyczne pobranie najnowszego ZIP
folder = Path("/content/weex_v645_results")
archiwa = list(folder.glob("weex_v645_*.zip"))
if not archiwa:
    raise FileNotFoundError("Nie znaleziono ZIP WEEX V6.45")
najnowszy = max(archiwa, key=lambda p: p.stat().st_mtime)
print("Pobieranie:", najnowszy.name)
files.download(str(najnowszy))
```
